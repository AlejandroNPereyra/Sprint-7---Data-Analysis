# Selección de Personal — Formulario y Clasificación

Herramienta en Python que actúa como formulario de inscripción y sistema de clasificación inicial para una oferta laboral interna.

**Reglas de negocio:**
- Solo se aceptan las **10 primeras personas válidas** que se inscriban.
- Cada candidato recibe una puntuación según experiencia, catalán, Excel y titulación.
- Se clasifica en: Prioridad Alta, Candidato Adecuado o No Prioritario.
- Al llegar a 10, se genera un informe final ordenado por prioridad y puntuación.

## Objetivo

Recoger datos de candidatos, validarlos, calcular su puntuación, clasificarlos y generar un informe final ordenado.

## Enfoque

1. **Recoger** los datos del candidato por teclado.
2. **Validar** que todos los campos son correctos.
3. **Calcular** la puntuación: `experiencia × 1.5 + catalán + Excel + titulación`.
4. **Clasificar** según los umbrales definidos.
5. **Mostrar** el resultado individual.
6. **Generar** un informe final ordenado por prioridad y puntuación.

## Restricción

Solo cuentan como inscritos los candidatos **válidos**. Un candidato inválido no ocupa plaza.

## Reglas de puntuación

| Concepto | Valor |
|---|---|
| Años de experiencia | × 1.5 |
| Catalán | 0–3 |
| Excel | 0–3 |
| Titulación | 0–3 |

**Fórmula:** `puntuación = experiencia × 1.5 + catalán + excel + titulación`

## Umbrales de clasificación

| Puntuación | Clasificación |
|---|---|
| ≥ 12 | Prioridad Alta |
| 8 ≤ puntuación < 12 | Candidato Adecuado |
| < 8 | No Prioritario |

## Recoger los datos del candidato

`obtener_datos()` solicita por teclado los 6 campos y los devuelve en un diccionario. Se aplica `.strip()` para eliminar espacios sobrantes.

````python
def obtener_datos():

    """
    Pide por teclado los datos de cada candidato

    Nombre y apellido
    Teléfono
    Años de experiencia
    Nivel de catalán (0–3)
    Excel (0–3)
    Nivel académico (0–3)

    .strip() elimina espacios sobrantes al principio y al final (útil si el usuario escribe " Laura " → "Laura").
    Devuelve un diccionario con esos datos en formato texto (todos son str, porque vienen de un input()).
    Por qué un diccionario: así luego se accede por nombre (datos["nombre"]) en vez de por posición, lo cual es más legible.
    """

    nombre = input("Nombre y apellido: ").strip()
    telefono = input("Teléfono de contacto: ").strip()
    experiencia = input("Años de experiencia en la empresa: ").strip()
    catalan = input("Nivel de catalán (0-3): ").strip()
    excel = input("Dominio de Excel (0-3): ").strip()
    titulo = input("Nivel académico (0-3): ").strip()

    return {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulo": titulo
    }
````

## Validar los datos

`validar_datos(datos)` comprueba cada campo:

| Campo | Regla |
|---|---|
| Nombre | No puede estar vacío |
| Teléfono | 9 dígitos numéricos |
| Experiencia | Entero ≥ 0 |
| Catalán, Excel, Titulación | Entero entre 0 y 3 |

Se usa `try/except` para capturar errores de conversión a entero. Si algo falla, devuelve `False` e imprime un mensaje.

````python
def validar_datos(datos):

    """
    Valida cada campo.
    Si algo falla, imprime un mensaje de error y devuelve False.
    Si todo está bien, devuelve True.

    Nombre:	No puede estar vacío
    Teléfono:	9 dígitos numéricos (isdigit() + len == 9)
    Experiencia:	Entero ≥ 0
    Niveles de catalán, Excel y título académico:	Enteros entre 0 y 3

    Detalle importante — try/except: cuando hacemos int(datos["experiencia"]),
    si el usuario escribe "hola" en vez de un número, Python devuelve ValueError
    y el programa se rompe. El try/except captura este error
    y lo convierte en un mensaje legible.

    Se valida aquí y no en obtener_datos para tener separación de responsabilidades.
    Una función recoge, y la otra valida. Así el código es más limpio y reutilizable.
    """

    # Nombre no vacío
    if not datos["nombre"]:
        print("Error: el nombre no puede estar vacío.")
        return False

    # Teléfono: 9 dígitos
    if not (datos["telefono"].isdigit() and len(datos["telefono"]) == 9):
        print("Error: el teléfono debe tener 9 dígitos.")
        return False

    # Experiencia: entero >= 0
    try:
        exp = int(datos["experiencia"])
        if exp < 0:
            raise ValueError
    except ValueError:
        print("Error: los años de experiencia deben ser un entero >= 0.")
        return False

    # Catalán, Excel, Título: enteros entre 0 y 3
    for campo in ["catalan", "excel", "titulo"]:
        try:
            valor = int(datos[campo])
            if valor < 0 or valor > 3:
                raise ValueError
        except ValueError:
            print(f"Error: {campo} debe ser un entero entre 0 y 3.")
            return False

    return True
````

## Calcular la puntuación

`calcular_puntuacion(datos)` aplica la fórmula del enunciado:

puntuación = experiencia × 1.5 + catalán + excel + titulación

Convierte los valores de texto a entero antes de operar.

````python
def calcular_puntuacion(datos):

    """
    Aplica la fórmula del enunciado:
    puntuación = (años_experiencia × 1.5) + catalán + excel + título

    Detalle: convierte los valores de texto a entero porque vienen de input() como str.
    Si no se convierte, experiencia * 1.5 daría error (no se puede multiplicar un string por un float).
    """

    experiencia = int(datos["experiencia"])
    catalan = int(datos["catalan"])
    excel = int(datos["excel"])
    titulo = int(datos["titulo"])
    return experiencia * 1.5 + catalan + excel + titulo
````

## Clasificar al candidato

`clasificar_candidato(puntuacion)` devuelve la etiqueta correspondiente según los umbrales:

- ≥ 12 → `"Prioridad Alta"`
- 8 ≤ puntuación < 12 → `"Candidato Adecuado"`
- < 8 → `"No Prioritario"`

El orden de los `if` importa: primero el umbral más alto.

````python
def clasificar_candidato(puntuacion):

    """
    Traduce la puntuación numérica a una etiqueta de texto según los umbrales de la empresa:


    Prioridad Alta ≥ 12
    Candidato Adecuado 8 – 11.99
    No Prioritario < 8

    Detalle: el orden de los if importa.
    Si pones primero >= 8, un 15 también entraría ahí.
    Al poner >= 12 primero, se descarta antes de llegar al elif.
    """

    if puntuacion >= 12:
        return "Prioridad Alta"
    elif puntuacion >= 8:
        return "Candidato Adecuado"
    else:
        return "No Prioritario"
````

## Mostrar el resultado individual

`mostrar_individual(datos, puntuacion, clasificacion)` imprime por pantalla los datos del candidato que acaba de registrarse, junto con su puntuación y clasificación.

````python
def mostrar_individual(datos, puntuacion, clasificacion):

    """
    Imprime por pantalla los datos del candidato que acaba de registrarse,
    junto con su puntuación y clasificación.

    El enunciado pide que "cada vez que alguien complete el formulario,
    se muestre su puntuación individual".
    """

    print("\n--- Candidato registrado ---")
    print(f"Nombre: {datos['nombre']}")
    print(f"Teléfono: {datos['telefono']}")
    print(f"Puntuación: {puntuacion}")
    print(f"Clasificación: {clasificacion}")
    print("----------------------------\n")
````

## Generar el informe final

`mostrar_informe(candidatos)` ordena la lista por:

1. **Prioridad** (Alta → Adecuado → No Prioritario), usando un diccionario `orden_prioridad`.
2. **Puntuación** descendente en caso de empate, usando `-puntuacion` como clave.

Después imprime una tabla con nombre, teléfono, puntuación y clasificación.

````python
def mostrar_informe(candidatos):

    """
    Define un diccionario orden_prioridad que traduce cada clasificación a un número
    (0 = más prioritario, 2 = menos).
    Esto es un truco para poder ordenar textos por un criterio lógico.

    Usa sorted() con una clave compuesta:

    Primero ordena por orden_prioridad[clasificacion] →
    Alta (0) antes que Adecuado (1) antes que No Prioritario (2).

    En caso de empate, ordena por -puntuacion →
    el signo negativo invierte el orden, así las puntuaciones más altas van primero.

    Imprime una tabla formateada con :<25, :<12, etc.
    (alineación a la izquierda con ancho fijo).

    Por qué -c["puntuacion"]: porque sorted() ordena de menor a mayor por defecto.
    Al negar, el 18 se convierte en -18 (menor) y va antes que -4.5.
    """

    # Ordenar: primero por prioridad (Alta > Adecuado > No Prioritario), después por puntuación desc
    orden_prioridad = {"Prioridad Alta": 0, "Candidato Adecuado": 1, "No Prioritario": 2}
    candidatos_ordenados = sorted(
        candidatos,
        key=lambda c: (orden_prioridad[c["clasificacion"]], -c["puntuacion"])
    )

    print("\n========== INFORME FINAL ==========")
    print(f"{'Nombre':<25} {'Teléfono':<12} {'Puntuación':<10} {'Clasificación'}")
    print("-" * 70)
    for c in candidatos_ordenados:
        print(f"{c['nombre']:<25} {c['telefono']:<12} {c['puntuacion']:<10} {c['clasificacion']}")
    print("===================================\n")
````

## Función principal

`main_seleccion()` orquesta todo el flujo:

1. Inicializa una lista vacía y el máximo de candidatos (10).
2. Repite el ciclo hasta tener 10 candidatos **válidos**:
   - Recoge datos.
   - Valida. Si falla, vuelve al inicio sin contar la inscripción.
   - Calcula puntuación y clasificación.
   - Añade el candidato a la lista.
   - Muestra el resultado individual.
3. Al terminar, genera el informe final.

**Detalle clave:** el `continue` tras la validación hace que solo los candidatos válidos ocupen plaza.

````python
def main_seleccion():

    """
    Inicializa una lista vacía candidatos y una constante MAX_CANDIDATOS = 10.

    Bucle while: se repite mientras haya menos de 10 candidatos válidos.

    En cada iteración:

    Muestra "Inscripción X de 10".
    Llama a obtener_datos() para recoger los datos.

    Llama a validar_datos():
    Si falla → mensaje "Vuelve a intentarlo" y continue → vuelve al inicio del bucle sin contar esa inscripción (importante: el candidato inválido no ocupa plaza).
    Si pasa → sigue.

    Calcula puntuación y clasificación.

    Crea un diccionario candidato con solo los campos que necesita el informe (nombre, teléfono, puntuación, clasificación).

    Lo añade a la lista.

    Muestra el resultado individual.

    Cuando la lista llega a 10 → sale del bucle y llama a mostrar_informe(candidatos).

    Detalle clave: el continue después de validar_datos hace que solo se cuenten como inscritos los candidatos válidos.
    Si alguien mete catalán = 7, no ocupa plaza.
    """

    candidatos = []
    MAX_CANDIDATOS = 10

    while len(candidatos) < MAX_CANDIDATOS:
        print(f"\nInscripción {len(candidatos) + 1} de {MAX_CANDIDATOS}")
        datos = obtener_datos()

        if not validar_datos(datos):
            print("Vuelve a intentarlo.\n")
            continue

        puntuacion = calcular_puntuacion(datos)
        clasificacion = clasificar_candidato(puntuacion)

        candidato = {
            "nombre": datos["nombre"],
            "telefono": datos["telefono"],
            "puntuacion": puntuacion,
            "clasificacion": clasificacion
        }
        candidatos.append(candidato)

        mostrar_individual(datos, puntuacion, clasificacion)

    mostrar_informe(candidatos)
````

## Ejecución

Ejecuta la función principal para iniciar el formulario.

````python
main_seleccion()
````

## Conclusiones

- El sistema acepta exactamente 10 candidatos **válidos**; los inválidos no ocupan plaza.
- La puntuación combina experiencia ponderada (× 1.5) con catalán, Excel y titulación.
- La clasificación usa tres umbrales: Alta (≥ 12), Adecuada (8–11.99), No Prioritario (< 8).
- El informe final ordena primero por prioridad y, en caso de empate, por puntuación descendente.
- El diseño es modular: cada función hace una sola cosa.
- La lógica de "solo cuentan los válidos" se resuelve con `continue` tras la validación.

`obtener_segmento_y_factor(codigo)` recibe un código (0–4) y devuelve tres cosas consultando la tabla:
- El segmento (texto legible)
- El factor numérico
- El mensaje publicitario

Sin condicionales: una sola línea de consulta.

````python
def obtener_segmento_y_factor(codigo):

    """
    Recibe un código de comportamiento (0-4) y devuelve:
    - segmento (str)
    - factor (float)
    - mensaje (str)
    Consulta directamente de la tabla.

    Devuelve una tupla con los tres valores.
    Si el código no existe, Python lanzaría KeyError
    (podríamos capturarlo, pero el enunciado garantiza códigos 0–4).
    """

    info = TABLA_COMPORTAMIENTO[codigo]
    return info["segmento"], info["factor"], info["mensaje"]
````

## Consultar la tabla de edad

`obtener_factor_edad(grupo_edad)` recibe un grupo de edad y devuelve su factor numérico. Consulta directa al diccionario.

````python
def obtener_factor_edad(grupo_edad):

    """
    Recibe un grupo de edad ('18-25', '26-40', '41-60', '60+')
    y devuelve su factor numérico.
    Sin condiciones: consulta directa a la tabla.
    Si el grupo no existe, KeyError. Los datos vienen garantizados.
    """

    return TABLA_EDAD[grupo_edad]
````

## Calcular el riesgo final

`calcular_riesgo(factor_codigo, factor_edad)` aplica la fórmula:

riesgo_final = factor_codigo × factor_edad

````python
def calcular_riesgo(factor_codigo, factor_edad):

    """
    Aplica la fórmula: riesgo_final = factor_codigo * factor_edad
    Es una multiplicación. No hay condición posible.
    Devuelve un float.
    """

    return factor_codigo * factor_edad
````

## Procesar todos los clientes

`procesar_clientes(clientes)` recorre la lista y, para cada cliente:
1. Consulta su segmento, factor y mensaje en la tabla de comportamiento.
2. Consulta su factor de edad en la tabla de edad.
3. Calcula el riesgo final.
4. Construye un diccionario ampliado.

Devuelve un **diccionario de diccionarios** indexado por `id_cliente`.

````python
def procesar_clientes(clientes):

    """
    Recibe la lista original de clientes y devuelve un diccionario de diccionarios
    indexado por id_cliente, con todos los campos ampliados:
    id_cliente, nombre, edad, grupo_edad, codigo_comportamiento,
    segmento, mensaje, factor_codigo, factor_edad, riesgo_final

    Itera sobre la lista original.

    Para cada cliente, llama a las tres funciones anteriores.
    Construye un diccionario ampliado.
    El resultado es un diccionario de diccionarios: clave = id_cliente, valor = diccionario del cliente.
    Cero condicionales.
    """

    resultado = {}

    for cliente in clientes:
        codigo = cliente["codi_comportament"]
        grupo = cliente["grup_edat"]

        # Consultas a las tablas (sin if)
        segmento, factor_codigo, mensaje = obtener_segmento_y_factor(codigo)
        factor_edad = obtener_factor_edad(grupo)

        # Cálculo del riesgo
        riesgo_final = calcular_riesgo(factor_codigo, factor_edad)

        # Construimos el diccionario ampliado del cliente
        resultado[cliente["id_client"]] = {
            "id_cliente":          cliente["id_client"],
            "nombre":              cliente["nom"],
            "edad":                cliente["edat"],
            "grupo_edad":          cliente["grup_edat"],
            "codigo_comportamiento": codigo,
            "segmento":            segmento,
            "mensaje":             mensaje,
            "factor_codigo":       factor_codigo,
            "factor_edad":         factor_edad,
            "riesgo_final":        riesgo_final
        }

    return resultado
````

## Mostrar la tabla procesada

`mostrar_diccionario(datos)` imprime por pantalla el diccionario de diccionarios en formato tabla alineada.

- `:<28` alinea a la izquierda con ancho 28.
- `.2f` formatea el riesgo con 2 decimales.

````python
def mostrar_diccionario(datos):

    """
    Muestra por pantalla el diccionario de diccionarios en formato tabla.

    Recorre el diccionario de diccionarios.
    Imprime cada cliente en una fila alineada.
    :<28 significa "alinear a la izquierda con ancho 28".
    .2f formatea el riesgo con 2 decimales.
    """

    print("\n" + "=" * 110)
    print(f"{'ID':<4} {'Nombre':<15} {'Edad':<5} {'Grupo':<7} {'Código':<7} "
          f"{'Segmento':<28} {'F.Cod':<7} {'F.Edad':<7} {'Riesgo':<6}")
    print("-" * 110)

    for id_cliente, c in datos.items():
        print(f"{c['id_cliente']:<4} {c['nombre']:<15} {c['edad']:<5} "
              f"{c['grupo_edad']:<7} {c['codigo_comportamiento']:<7} "
              f"{c['segmento']:<28} {c['factor_codigo']:<7} "
              f"{c['factor_edad']:<7} {c['riesgo_final']:<6.2f}")

    print("=" * 110 + "\n")
````

## Responder las 4 preguntas de marketing

`mostrar_respuestas(datos)` resuelve las preguntas **sin condicionales**, usando ordenación, agrupación, filtrado y máximo.

| Pregunta | Técnica sin `if` |
|---|---|
| 1. ¿A quién contactar primero? | `sorted(..., reverse=True)[:3]` |
| 2. ¿Mismo comportamiento, distinta edad? | Agrupación con `setdefault` |
| 3. ¿A quién NO contactar? | List comprehension con filtro `riesgo_final < 0.20` |
| 4. ¿Cliente más rentable? | `max(..., key=lambda c: c["riesgo_final"])` |

> El filtro `riesgo_final < 0.20` es una expresión booleana dentro de una comprensión de listas, no un `if/elif/else` de control de flujo.

````python
def mostrar_respuestas(datos):

    """
    Responde las 4 preguntas de marketing a partir del diccionario procesado.

    Pregunta                                      Como resolvemos sin condicional

    1. Contactar primero	                        sorted (..., reverse=True)[:3]
    2. Mismo comportamiento, distinta edad.      	grupos.setdefault(codigo, []).append(c) (agrupación)
    3. Poco interés	                              List comprehension con filtro riesgo_final < 0.20
    4. Más rentable	                              max(..., key=lambda c: c["riesgo_final"])

    Nota: el filtro riesgo_final < 0.20 es una expresión booleana
    dentro de una comprensión de listas, no un if/elif/else de control de flujo.

    El enunciado prohíbe los condicionales de segmentación, no las expresiones de filtrado.
    """

    # Convertimos a lista para poder ordenar y filtrar
    lista = list(datos.values())

    # ------------------------------------------------------------
    # PREGUNTA 1: ¿A quién contactar primero?
    # Ordenamos por riesgo_final descendente y mostramos los 3 primeros.
    # ------------------------------------------------------------

    print("\n" + "=" * 60)
    print("PREGUNTA 1: Clientes a contactar PRIMERO (top 3 por riesgo)")
    print("=" * 60)
    top3 = sorted(lista, key=lambda c: c["riesgo_final"], reverse=True)[:3]
    for c in top3:
        print(f"  - {c['nombre']} (ID {c['id_cliente']}) | "
              f"Segmento: {c['segmento']} | Riesgo: {c['riesgo_final']:.2f}")

    # ------------------------------------------------------------
    # PREGUNTA 2: Diferenciar clientes con el mismo comportamiento
    # pero con edades diferentes.
    # Agrupamos por el codigo_comportamiento y mostramos los riesgos de cada grupo.
    # ------------------------------------------------------------

    print("\n" + "=" * 60)
    print("PREGUNTA 2: Mismo comportamiento, edades diferentes")
    print("=" * 60)

    grupos = {}
    for c in lista:
        grupos.setdefault(c["codigo_comportamiento"], []).append(c)

    for codigo in sorted(grupos.keys()):
        print(f"\n  Código {codigo} ({grupos[codigo][0]['segmento']}):")
        for c in grupos[codigo]:
            print(f"    - {c['nombre']:<15} Edad {c['edad']:<3} "
                  f"Grupo {c['grupo_edad']:<6} → Riesgo: {c['riesgo_final']:.2f}")

    # ------------------------------------------------------------
    # PREGUNTA 3: Clientes con poco interés (en los que no vale la pena invertir)
    # Filtrado por riesgo_final por debajo de un umbral.
    # ------------------------------------------------------------

    print("\n" + "=" * 60)
    print("PREGUNTA 3: Clientes con poco interés (riesgo < 0.20)")
    print("=" * 60)
    poco_interes = [c for c in lista if c["riesgo_final"] < 0.20]
    for c in poco_interes:
        print(f"  - {c['nombre']} (ID {c['id_cliente']}) | "
              f"Segmento: {c['segmento']} | Riesgo: {c['riesgo_final']:.2f}")

    # ------------------------------------------------------------
    # PREGUNTA 4: Cliente más rentables de recuperar
    # Los de riesgo_final más alto.
    # ------------------------------------------------------------

    print("\n" + "=" * 60)
    print("PREGUNTA 4: Cliente MÁS RENTABLE de recuperar")
    print("=" * 60)
    mejor = max(lista, key=lambda c: c["riesgo_final"])
    print(f"  - {mejor['nombre']} (ID {mejor['id_cliente']})")
    print(f"    Segmento: {mejor['segmento']}")
    print(f"    Mensaje recomendado: {mejor['mensaje']}")
    print(f"    Riesgo final: {mejor['riesgo_final']:.2f}")
    print("=" * 60 + "\n")
````

## Función principal

`main_marketing()` orquesta todo:

````python
def main_marketing():

    """
    Función principal: procesa los clientes, muestra la tabla
    y responde a las 4 preguntas de marketing.

    # 1. Procesar
    datos = procesar_clientes(clientes)

    # 2. Mostrar tabla
    mostrar_diccionario(datos)

    # 3. Responder preguntas
    mostrar_respuestas(datos)
    """

    # Datos originales
    clientes = [
        {"id_client": 1,  "nom": "Laura Serra",  "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
        {"id_client": 2,  "nom": "Marc Vidal",   "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
        {"id_client": 3,  "nom": "Ana López",    "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
        {"id_client": 4,  "nom": "Joan Riera",   "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
        {"id_client": 5,  "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
        {"id_client": 6,  "nom": "Luis García",  "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
        {"id_client": 7,  "nom": "Pilar Sánchez","edat": 61, "grup_edat": "60+",   "codi_comportament": 3},
        {"id_client": 8,  "nom": "Eva Martín",   "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
        {"id_client": 9,  "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
        {"id_client": 10, "nom": "Núria Costa",  "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
    ]

    # 1. Procesar
    datos = procesar_clientes(clientes)

    # 2. Mostrar tabla
    mostrar_diccionario(datos)

    # 3. Responder preguntas
    mostrar_respuestas(datos)
````

# Gestión de riesgo y priorización de pacientes

Trabajas como analista de datos en un centro médico especializado. Varios pacientes han enviado sus síntomas a través de un formulario online, pero los datos llegan en un formato desordenado y poco estructurado.

**Objetivo:** transformar esos datos en información útil para que los médicos prioricen el orden de atención según la gravedad.

**Enfoque:** diseño modular, cada función hace una sola cosa.

## Objetivo

Procesar una lista de strings brutos con información de pacientes y producir:

- Nombre
- Edad
- Síntomas normalizados
- Riesgo total

Y un ranking de prioridad (más riesgo → primero).

## Enfoque

1. **Recibir** los datos brutos (strings con formato `"Nombre|Edad|síntomas"`).
2. **Parsear** cada línea para separar nombre, edad y síntomas.
3. **Limpiar y normalizar** los síntomas (minúsculas, sin espacios, guion bajo).
4. **Calcular** el riesgo total sumando los valores del diccionario `niveles_riesgo`.
5. **Ordenar** por riesgo descendente (y edad en caso de empate).
6. **Mostrar** la tabla completa y el ranking.

## Restricción

El enunciado no da la lista de funciones: hay que diseñarla. Se aplica el principio de **una función = una responsabilidad**.

## Datos de entrada

Cada paciente llega como un string con este formato:

"Maria|45|tos, Fiebre Alta, dolor_pecho"


Separado por `|` en tres partes:
1. Nombre
2. Edad
3. Lista de síntomas (separados por comas, con espacios, mayúsculas mezcladas…)

El diccionario `niveles_riesgo` asocia cada síntoma normalizado con su valor de riesgo.

````python
# ============================================================
#                       DATOS DE ENTRADA
# ============================================================

pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
````

## ¿Qué funciones necesito?

Como no hay lista, la diseñamos. Aplicamos el principio de **una función = una responsabilidad**:

| Función | Responsabilidad |
|---|---|
| `normalizar_sintoma(sintoma)` | Convertir "Fiebre Alta" → "fiebre_alta" |
| `limpiar_sintomas(texto)` | Recibir un string de síntomas y devolver una lista normalizada |
| `parsear_paciente(linea)` | Convertir un string bruto en diccionario limpio |
| `calcular_riesgo_total(sintomas, niveles_riesgo)` | Sumar los riesgos de todos los síntomas |
| `procesar_pacientes(pacientes, niveles_riesgo)` | Recorrer la lista y aplicar todo lo anterior |
| `ordenar_por_riesgo(datos)` | Ordenar de mayor a menor riesgo |
| `mostrar_resultados(datos)` | Visualizar tabla + ranking |
| `main_triage()` | Orquestar todas las anteriores |

## ¿Cómo normalizo un síntoma?

Los síntomas vienen sucios:

| Original | Problema | Normalizado |
|---|---|---|
| `tos` | ya está bien | `tos` |
| `Fiebre Alta` | mayúsculas y espacio | `fiebre_alta` |
| `dolor_pecho` | ya está bien | `dolor_pecho` |
| `dificultad RESPIRAR` | mayúsculas y espacio | `dificultad_respirar` |
| `fatiga ` | espacio al final | `fatiga` |

**Regla de normalización:**
1. Convertir a minúsculas.
2. Eliminar espacios al principio y al final.
3. Reemplazar espacios internos por guion bajo `_`.

Así `"Fiebre Alta"` → `"fiebre_alta"`, que coincide con la clave del diccionario `niveles_riesgo`.

````python
def normalizar_sintoma(sintoma):

    """
    Convierte un síntoma a formato estándar:
    - minúsculas
    - sin espacios al principio/final
    - espacios internos → guion bajo
    Ejemplo: " Fiebre Alta " → "fiebre_alta"
    """

    return sintoma.strip().lower().replace(" ", "_")
````

````python
def limpiar_sintomas(texto):

    """
    Recibe un string con síntomas separados por comas
    y devuelve una lista de síntomas normalizados.
    Ejemplo: "tos, Fiebre Alta, dolor_pecho" → ["tos", "fiebre_alta", "dolor_pecho"]
    """

    partes = texto.split(",")
    sintomas = [normalizar_sintoma(p) for p in partes]
    # Descartamos los vacíos (por si hay comas dobles o espacios)
    return [s for s in sintomas if s]
````

## ¿Cómo separo los síntomas y parseo la línea?

El string de síntomas viene separado por comas, pero puede haber espacios variables.

**Regla:**
1. Dividir por coma.
2. Limpiar cada trozo (quitar espacios).
3. Normalizar cada trozo.
4. Descartar los vacíos.

El resultado es una lista limpia: `["tos", "fiebre_alta", "dolor_pecho"]`.

`parsear_paciente(linea)` convierte una línea bruta en un diccionario limpio con `nombre`, `edad` y `sintomas`.

````python
def parsear_paciente(linea):

    """
    Convierte una línea bruta en un diccionario limpio.
    Formato de entrada: "Nombre|Edad|síntomas"
    Devuelve: {"nombre":..., "edad":..., "sintomas":[...]}
    """

    partes = linea.split("|")
    nombre = partes[0].strip()
    edad = int(partes[1].strip())
    sintomas = limpiar_sintomas(partes[2])
    return {
        "nombre": nombre,
        "edad": edad,
        "sintomas": sintomas
    }
````

## ¿Cómo calculo el riesgo total?

Para cada síntoma de la lista, buscamos su valor en `niveles_riesgo` y los sumamos.

**Detalle importante:** ¿qué pasa si un síntoma no está en el diccionario?

| Opción | Comportamiento |
|---|---|
| A | Ignorarlo (riesgo 0) |
| B | Lanzar error |
| C | Asignarle un valor por defecto |

Lo más robusto para datos reales es **ignorarlo con un valor por defecto de 0**, usando `.get(sintoma, 0)`. Así el programa no se rompe si llega un síntoma nuevo.

````python
def calcular_riesgo_total(sintomas, niveles_riesgo):

    """
    Suma el riesgo de cada síntoma según el diccionario niveles_riesgo.
    Si un síntoma no está en el diccionario, se le asigna 0 (tolerancia a datos nuevos).
    """

    return sum(niveles_riesgo.get(s, 0) for s in sintomas)
````

````python
def procesar_pacientes(pacientes, niveles_riesgo):

    """
    Recorre la lista de strings brutos y devuelve una lista de diccionarios
    con nombre, edad, síntomas normalizados y riesgo_total.
    """

    resultado = []
    for linea in pacientes:
        paciente = parsear_paciente(linea)
        paciente["riesgo_total"] = calcular_riesgo_total(
            paciente["sintomas"], niveles_riesgo
        )
        resultado.append(paciente)
    return resultado
````

## Paso 6: ¿Cómo ordeno por prioridad?

Ordenar por `riesgo_total` de mayor a menor. Para eso se usa `sorted()` con `reverse=True` y una clave que apunte al campo `riesgo_total`.

En caso de empate, se añade un segundo criterio: por **edad descendente** (los mayores primero), que es una decisión clínica razonable.

````python
def ordenar_por_riesgo(datos):
    """
    Ordena los pacientes por riesgo_total descendente.
    En caso de empate, ordena por edad descendente (criterio clínico).
    """
    return sorted(
        datos,
        key=lambda p: (p["riesgo_total"], p["edad"]),
        reverse=True
    )
````

## Paso 7: ¿Cómo lo muestro?

Dos salidas:

1. **Tabla completa** con todos los campos (nombre, edad, síntomas, riesgo).
2. **Ranking de prioridad** ordenado, con el orden de atención.

````python
def mostrar_resultados(datos):
    """
    Muestra la tabla completa de pacientes y el ranking de prioridad.
    """
    # --- Tabla completa ---
    print("\n" + "=" * 90)
    print(f"{'Nombre':<10} {'Edad':<6} {'Síntomas':<45} {'Riesgo':<6}")
    print("-" * 90)
    for p in datos:
        sintomas_str = ", ".join(p["sintomas"])
        print(f"{p['nombre']:<10} {p['edad']:<6} {sintomas_str:<45} {p['riesgo_total']:<6}")
    print("=" * 90)

    # --- Ranking de prioridad ---
    print("\n" + "=" * 50)
    print("RANKING DE PRIORIDAD (más riesgo → primero)")
    print("=" * 50)
    for i, p in enumerate(ordenar_por_riesgo(datos), start=1):
        print(f"  {i}. {p['nombre']:<10} Edad {p['edad']:<3} "
              f"→ Riesgo: {p['riesgo_total']}")
    print("=" * 50 + "\n")
````

## Paso 8: Función principal

`main_triage()` orquesta todo el flujo:

1. Procesa los datos brutos.
2. Muestra la tabla completa y el ranking de prioridad.

````python
def main_triage():

    """
    Función principal: procesa, ordena y muestra los resultados.
    """

    # 1. Procesar los datos brutos
    datos = procesar_pacientes(pacientes, niveles_riesgo)

    # 2. Mostrar tabla y ranking
    mostrar_resultados(datos)
````

## Paso 9: Ejecución

Ejecuta la función principal para procesar los pacientes y ver los resultados.

````python
main_triage()
````

## Enfoque

|---|---|---|---|
| 0 | Bajo riesgo | 120 | 0.01 |

| 3 | Riesgo crítico | 500 | 0.30 |

## Fórmula del precio final

precio_final = precio_base × (1 + edad / 100)

La edad se divide por 100 para convertirla en un factor decimal.

## Datos de entrada

Tenemos una lista de clientes, cada uno con `nombre`, `edad` y `codigo`.

Y una tabla `niveles_seguro` que asocia cada código con su categoría, precio base y probabilidad de fraude.

````python
# ============================================================
# DATOS DE ENTRADA
# ============================================================

clientes = [
    {"nombre": "Ana",    "edad": 42, "codigo": 0},
    {"nombre": "Carlos", "edad": 30, "codigo": 2},
    {"nombre": "Isabel", "edad": 55, "codigo": 3},
    {"nombre": "Jorge",  "edad": 40, "codigo": 1},
    {"nombre": "Marta",  "edad": 28, "codigo": 0}
]

niveles_seguro = {
    0: {"categoria": "Bajo riesgo",    "precio_base": 120, "fraude": 0.01},
    1: {"categoria": "Riesgo medio",   "precio_base": 200, "fraude": 0.05},
    2: {"categoria": "Riesgo alto",    "precio_base": 350, "fraude": 0.15},
    3: {"categoria": "Riesgo crítico", "precio_base": 500, "fraude": 0.30}
}
````

## ¿Qué funciones necesito?

## ¿Cómo clasifico sin `if/elif`?

Uso la tabla `niveles_seguro` como diccionario de consulta. La clave es el código (0–3) y el valor es otro diccionario con categoría, precio base y fraude.

No hay ninguna pregunta condicional: solo consultas por clave. Añadir un código nuevo = añadir una entrada al diccionario, sin tocar la lógica.

`obtener_categoria(codigo, niveles_seguro)` devuelve los tres valores consultando la tabla.

````python
def obtener_categoria(codigo, niveles_seguro):

    """
    Consulta la tabla niveles_seguro y devuelve:

    - categoria (str)
    - precio_base (int)
    - fraude (float)
    """

    info = niveles_seguro[codigo]
    return info["categoria"], info["precio_base"], info["fraude"]
````

## ¿Cómo calculo el precio final?

La fórmula del enunciado es directa:

precio_final = precio_base × (1 + edad/100)

No hay condición. Solo una multiplicación. La edad se divide por 100 para convertirla en un factor decimal.

````python
def calcular_precio_final(precio_base, edad):

    """
    Aplica la fórmula: precio_final = precio_base * (1 + edad/100)
    """

    return precio_base * (1 + edad / 100)
````

## ¿Qué estructura final tendrá cada cliente?

Cada cliente procesado será un diccionario con:

- `nombre`
- `edad`
- `codigo`
- `categoria`
- `precio_base`
- `precio_final`
- `probabilidad_fraude`

Y todos juntos formarán una lista de diccionarios (o un diccionario indexado por nombre).

- `procesar_cliente` amplía un cliente con todos los campos nuevos.
- `procesar_clientes` recorre la lista completa y aplica lo anterior.

````python
def procesar_cliente(cliente, niveles_seguro):

    """
    Amplía un cliente con categoría, precio_base, precio_final y fraude.
    Devuelve un diccionario nuevo con todos los campos.
    """

    codigo = cliente["codigo"]
    edad = cliente["edad"]

    categoria, precio_base, fraude = obtener_categoria(codigo, niveles_seguro)
    precio_final = calcular_precio_final(precio_base, edad)

    return {
        "nombre":              cliente["nombre"],
        "edad":                edad,
        "codigo":              codigo,
        "categoria":           categoria,
        "precio_base":         precio_base,
        "precio_final":        precio_final,
        "probabilidad_fraude": fraude
    }
````

````python
def procesar_clientes(clientes, niveles_seguro):

    """
    Recorre la lista de clientes y devuelve una lista de diccionarios ampliados.
    """

    return [procesar_cliente(c, niveles_seguro) for c in clientes]
````

## ¿Cómo ordeno?

El enunciado pide dos ordenaciones:

1. **Por probabilidad de fraude descendente** → para saber a quién revisar primero.
2. **Por precio final descendente** → para saber quién paga más.

Se resuelve con `sorted()` y una `key` que apunte al campo correspondiente, con `reverse=True` para ordenar de mayor a menor.

En caso de empate en fraude, se añade un segundo criterio por `precio_final` descendente.

````python
def ordenar_por_fraude(datos):

    """
    Ordena los clientes por probabilidad de fraude descendente.
    En caso de empate, por precio_final descendente.
    """

    return sorted(
        datos,
        key=lambda c: (c["probabilidad_fraude"], c["precio_final"]),
        reverse=True
    )
````

````python
def ordenar_por_precio(datos):

    """
    Ordena los clientes por precio_final descendente.
    """

    return sorted(
        datos,
        key=lambda c: c["precio_final"],
        reverse=True
    )
````

## ¿Cómo lo presento?

Dos salidas:

1. **Tabla completa** con todos los campos.
2. **Análisis de negocio** con las 3 respuestas.

`mostrar_tabla(datos)` imprime la tabla alineada con `:<10`, `:<16`, etc.

````python
def mostrar_tabla(datos):

    """
    Muestra por pantalla la tabla completa de clientes procesados.
    """

    print("\n" + "=" * 100)
    print(f"{'Nombre':<10} {'Edad':<6} {'Categoría':<16} {'P.Base':<8} "
          f"{'P.Final':<10} {'Fraude':<8}")
    print("-" * 100)
    for c in datos:
        print(f"{c['nombre']:<10} {c['edad']:<6} {c['categoria']:<16} "
              f"{c['precio_base']:<8} {c['precio_final']:<10.2f} "
              f"{c['probabilidad_fraude']:<8.2f}")
    print("=" * 100 + "\n")
````

## ¿Cómo respondo a las 3 preguntas?

| Pregunta | Técnica sin `if` |
|---|---|
| 1. ¿Cliente con más probabilidad de fraude? | `max(..., key=lambda c: c["probabilidad_fraude"])` |
| 2. ¿Cliente con precio final más alto? | `max(..., key=lambda c: c["precio_final"])` |
| 3. ¿Prioritario de revisar manualmente? | El de mayor fraude (o combinación fraude + precio) |

En todos los casos se usan **operaciones de ordenación y máximo**, no condicionales.

````python
def mostrar_analisis(datos):

    """
    Responde a las 3 preguntas de negocio:
    1. Cliente con más probabilidad de fraude.
    2. Cliente con precio final más alto.
    3. Cliente prioritario de revisar manualmente.
    """

    # ------------------------------------------------------------
    # PREGUNTA 1: Cliente con más probabilidad de fraude
    # ------------------------------------------------------------

    print("\n" + "=" * 60)
    print("ANÁLISIS DE NEGOCIO")
    print("=" * 60)

    mas_fraude = max(datos, key=lambda c: c["probabilidad_fraude"])
    print(f"\n1. Cliente con MÁS probabilidad de fraude:")
    print(f"   → {mas_fraude['nombre']} (Edad {mas_fraude['edad']}) | "
          f"Categoría: {mas_fraude['categoria']} | "
          f"Fraude: {mas_fraude['probabilidad_fraude']:.2f}")

    # ------------------------------------------------------------
    # PREGUNTA 2: Cliente con precio final más alto
    # ------------------------------------------------------------

    mas_caro = max(datos, key=lambda c: c["precio_final"])
    print(f"\n2. Cliente con PRECIO FINAL más alto:")
    print(f"   → {mas_caro['nombre']} (Edad {mas_caro['edad']}) | "
          f"Categoría: {mas_caro['categoria']} | "
          f"Precio final: {mas_caro['precio_final']:.2f}")

    # ------------------------------------------------------------
    # PREGUNTA 3: Cliente prioritario de revisar manualmente
    # Criterio: el de mayor fraude (ya calculado). Si hay empate,
    # se puede combinar con precio_final.
    # ------------------------------------------------------------

    print(f"\n3. Cliente PRIORITARIO de revisar manualmente:")
    print(f"   → {mas_fraude['nombre']} "
          f"(mayor probabilidad de fraude: {mas_fraude['probabilidad_fraude']:.2f})")

    print("\n" + "=" * 60 + "\n")
````

## Función principal

`main_seguros()` orquesta todo:

1. Procesa los clientes.
2. Muestra la tabla completa.
3. Muestra el orden por fraude.
4. Muestra el orden por precio final.
5. Muestra el análisis de negocio.

````python
def main_seguros():

    """
    Función orquestadora: procesa, ordena y muestra los resultados.
    """

    # 1. Procesar clientes

    datos = procesar_clientes(clientes, niveles_seguro)

    # 2. Mostrar tabla completa

    mostrar_tabla(datos)

    # 3. Mostrar orden por fraude

    print("ORDEN POR PROBABILIDAD DE FRAUDE (descendente):")
    for c in ordenar_por_fraude(datos):
        print(f"  - {c['nombre']:<10} Fraude: {c['probabilidad_fraude']:.2f} | "
              f"Precio final: {c['precio_final']:.2f}")

    # 4. Mostrar orden por precio final

    print("\nORDEN POR PRECIO FINAL (descendente):")
    for c in ordenar_por_precio(datos):
        print(f"  - {c['nombre']:<10} Precio final: {c['precio_final']:.2f} | "
              f"Fraude: {c['probabilidad_fraude']:.2f}")

    # 5. Análisis de negocio

    mostrar_analisis(datos)
````

**¿Cómo construyo los conjuntos de caracteres?**

Necesito cuatro conjuntos, uno por tipo:

Mayúsculas	A–Z

Minúsculas	a–z

Números	0–9

Signos	!@#$%^&*()-_=+[]{};:,.<>?

Se pueden construir con string.ascii_uppercase, string.ascii_lowercase, string.digits y una cadena propia de signos.

**Importante:** el conjunto total es la unión de los tipos activos. Los obligatorios son uno por cada tipo activo.

---

````python
import numpy as np
import string

# ============================================================
# CONJUNTOS DE CARACTERES POR TIPO
# ============================================================

CONJUNTO_MAYUSCULAS = string.ascii_uppercase          # A-Z
CONJUNTO_MINUSCULAS = string.ascii_lowercase          # a-z
CONJUNTO_NUMEROS    = string.digits                   # 0-9
CONJUNTO_SIGNOS     = "!@#$%^&*()-_=+[]{};:,.<>?"
````

````python
def construir_conjuntos(mayusculas, minusculas, numeros, signos):

    """
    Devuelve dos cosas:
    - conjunto_total: string con todos los caracteres disponibles
    - conjuntos_activos: lista de strings, uno por cada tipo activo
    Sin if/elif: usamos una lista de tuplas (flag, conjunto) y filtramos.
    """

    # Tabla de (flag, conjunto) — sustituye a los if
    tabla = [
        (mayusculas, CONJUNTO_MAYUSCULAS),
        (minusculas, CONJUNTO_MINUSCULAS),
        (numeros,    CONJUNTO_NUMEROS),
        (signos,     CONJUNTO_SIGNOS)
    ]

    # Filtramos los activos con una comprensión de listas (no es if/elif de control)
    conjuntos_activos = [conj for flag, conj in tabla if flag]
    conjunto_total = "".join(conjuntos_activos)

    return conjunto_total, conjuntos_activos
````

**¿Qué pasa si la longitud es menor que el número de tipos?**

Ejemplo: pides longitud 2 pero quieres mayúsculas, minúsculas, números y signos (4 tipos). Es imposible meter al menos uno de cada tipo en 2 caracteres.

Solución: validar antes de generar. Si longitud < número de tipos activos, se lanza un error o se devuelve un mensaje. Lo más limpio es lanzar una excepción (ValueError) con un mensaje claro.

````python
def validar_longitud(longitud, conjuntos_activos):

    """
    Comprueba que la longitud sea suficiente para incluir al menos
    un carácter de cada tipo activo.
    Lanza ValueError si no es posible.
    """

    if longitud < len(conjuntos_activos):
        raise ValueError(
            f"Longitud insuficiente: se piden {len(conjuntos_activos)} tipos "
            f"pero solo hay {longitud} caracteres."
        )
````

---

**¿Cómo garantizo que haya al menos uno de cada tipo?**
El truco es en dos fases:

Fase obligatoria: elijo un carácter de cada tipo activo. Esto garantiza el mínimo.

Fase de relleno: completo hasta la longitud deseada con caracteres aleatorios del conjunto total (que puede repetir tipos).

Después mezclo todo para que los obligatorios no queden al principio.

Ejemplo con longitud 12 y los 3 tipos activos (mayús, minús, núm):

Obligatorios: A, x, 7 (3 caracteres)

Relleno: 9 caracteres aleatorios del conjunto total

Total: 12 caracteres, mezclados

---

---

**¿Cómo uso NumPy para aleatorizar?**

El módulo random de NumPy tiene funciones equivalentes a las de Python estándar

random.choice(lista)	  -          np.random.choice(lista)

random.shuffle(lista)	 -           np.random.shuffle(lista)

random.choices(lista, k=n)  -      np.random.choice(lista, size=n)

**Detalle:** np.random.choice devuelve un ndarray, no una lista.

Hay que convertirlo con .tolist() si queremos operar como lista.

---

````python
def generar_obligatorios(conjuntos_activos):

    """
    Elige un carácter aleatorio de cada tipo activo.
    Devuelve una lista con un carácter por tipo.
    """

    return [np.random.choice(list(conj)) for conj in conjuntos_activos]
````

````python
def generar_resto(conjunto_total, cantidad):

    """
    Elige 'cantidad' caracteres aleatorios del conjunto total.
    Devuelve una lista.
    """

    if cantidad <= 0:
        return []
    return np.random.choice(list(conjunto_total), size=cantidad).tolist()
````

---

**¿Cómo mezclo sin perder caracteres?**

np.random.shuffle modifica la lista in-place y devuelve None.

Así que hay que:

Convertir la lista a un array de NumPy o pasarla directamente si es lista.

Llamar a np.random.shuffle(lista).

La lista queda modificada.

---

````python
def mezclar_caracteres(lista_caracteres):

    """
    Baraja la lista in-place con NumPy y la devuelve.
    """

    np.random.shuffle(lista_caracteres)
    return lista_caracteres
````

````python
def generar_password(longitud=12,
                     mayusculas=True,
                     minusculas=True,
                     numeros=True,
                     signos=False):

    """
    Genera una contraseña segura con los parámetros indicados.
    Garantiza al menos un carácter de cada tipo activo.

    Parámetros por defecto (obligatorios del diseño):
    - mayusculas=True
    - minusculas=True
    - numeros=True
    - signos=False
    """

    # 1. Construir conjuntos

    conjunto_total, conjuntos_activos = construir_conjuntos(
        mayusculas, minusculas, numeros, signos
    )

    # 2. Validar longitud

    validar_longitud(longitud, conjuntos_activos)

    # 3. Generar obligatorios (uno por tipo activo)

    obligatorios = generar_obligatorios(conjuntos_activos)

    # 4. Generar el resto

    restantes = generar_resto(conjunto_total, longitud - len(obligatorios))

    # 5. Unir y mezclar

    todos = obligatorios + restantes
    mezclados = mezclar_caracteres(todos)

    # 6. Unir en string

    return "".join(mezclados)
````

---

**¿Cómo lo pruebo?**

## Tabla de consulta (sin `if/elif`)

En lugar de usar condicionales para decidir si un partido es victoria, empate o derrota, usamos el **signo de la diferencia de goles** como clave de una tabla.

- `signo = 1` → victoria → +3 puntos
- `signo = 0` → empate → +1 punto
- `signo = -1` → derrota → +0 puntos

El signo se calcula con una expresión booleana sin ramas:
`signo = (gf > gc) - (gf < gc)`

````python
# ============================================================
#                     CONSTANTES Y TABLAS
# ============================================================

TABLA_SIGNO = {
    1:  {"guanyats": 1, "empatats": 0, "perduts": 0, "punts": 3},
    0:  {"guanyats": 0, "empatats": 1, "perduts": 0, "punts": 1},
    -1: {"guanyats": 0, "empatats": 0, "perduts": 1, "punts": 0}
}
````

## Lectura y parseo

- `leer_archivo(ruta)`: lee el archivo y devuelve una lista de líneas no vacías.
- `parsear_partido(linea)`: convierte una línea en una tupla `(local, goles_local, visitante, goles_visitante)`.

El archivo usa **tabuladores** como separador y un **guion** dentro del marcador: Manlleu 0-1 Granollers


````python
def leer_archivo(ruta):

    """
    Lee el archivo de partidos y devuelve una lista de líneas no vacías.
    """

    with open(ruta, "r", encoding="utf-8") as f:
        lineas = [l.strip() for l in f if l.strip()]
    return lineas
````

````python
def parsear_partido(linea):

    """
    Convierte una línea en una tupla:
    (equipo_local, goles_local, equipo_visitante, goles_visitante)

    Formato del archivo: "EquipoLocal\tGolesLocal-GolesVisitante\tEquipoVisitante"
    Ejemplo: "Manlleu\t0-1\tGranollers"
    """

    partes = linea.split("\t")

    local = partes[0].strip()
    marcador = partes[1].strip()          # "0-1"
    visitante = partes[2].strip()

    goles_local, goles_visitante = marcador.split("-")
    goles_local = int(goles_local)
    goles_visitante = int(goles_visitante)

    return local, goles_local, visitante, goles_visitante
````

## Acumulación de estadísticas

- `inicializar_equipo(nombre)`: crea el diccionario base de un equipo con todos los contadores a 0.
- `actualizar_equipo(equipo, gf, gc)`: suma goles y actualiza victorias/empates/derrotas y puntos usando `TABLA_SIGNO`.
- `procesar_partidos(lineas)`: recorre todas las líneas y acumula estadísticas en un diccionario de diccionarios.

````python
def inicializar_equipo(nombre):

    """
    Devuelve el diccionario base de un equipo con todos los contadores a 0.
    """

    return {
        "equip":       nombre,
        "gols_favor":  0,
        "gols_contra": 0,
        "guanyats":    0,
        "empatats":    0,
        "perduts":     0,
        "punts":       0
    }
````

````python
def actualizar_equipo(equipo, goles_favor, goles_contra):

    """
    Actualiza las estadísticas de un equipo con un partido.
    Sin if/elif: el signo de la diferencia es la clave de TABLA_SIGNO.
    """

    equipo["gols_favor"]  += goles_favor
    equipo["gols_contra"] += goles_contra

    signo = (goles_favor > goles_contra) - (goles_favor < goles_contra)
    incrementos = TABLA_SIGNO[signo]

    equipo["guanyats"] += incrementos["guanyats"]
    equipo["empatats"] += incrementos["empatats"]
    equipo["perduts"]  += incrementos["perduts"]
    equipo["punts"]    += incrementos["punts"]
````

````python
def procesar_partidos(lineas):

    """
    Recorre todas las líneas, parsea cada partido y acumula estadísticas.
    Devuelve un diccionario de diccionarios indexado por nombre de equipo.
    """

    estadisticas = {}

    for linea in lineas:
        local, goles_local, visitante, goles_visitante = parsear_partido(linea)

        for nombre in (local, visitante):
            if nombre not in estadisticas:
                estadisticas[nombre] = inicializar_equipo(nombre)

        actualizar_equipo(estadisticas[local], goles_local, goles_visitante)
        actualizar_equipo(estadisticas[visitante], goles_visitante, goles_local)

    return estadisticas
````

## Ordenación y tabla final

- `ordenar_por_puntos(estadisticas)`: convierte el dict en lista de registros y la ordena por puntos descendente (y diferencia de goles en caso de empate).
- `construir_tabla(registros)`: transpone la lista de registros a un diccionario de listas.

````python
def ordenar_por_puntos(estadisticas):

    """
    Convierte el diccionario de estadísticas en una lista de registros ordenada
    por puntos descendente. En caso de empate, por diferencia de goles.
    """

    registros = list(estadisticas.values())
    return sorted(
        registros,
        key=lambda r: (r["punts"], r["gols_favor"] - r["gols_contra"]),
        reverse=True
    )
````

````python
def construir_tabla(registros):

    """
    Convierte la lista ordenada de registros en un diccionario de listas.
    """

    return {
        "equip":       [r["equip"]       for r in registros],
        "gols_favor":  [r["gols_favor"]  for r in registros],
        "gols_contra": [r["gols_contra"] for r in registros],
        "guanyats":    [r["guanyats"]    for r in registros],
        "empatats":    [r["empatats"]    for r in registros],
        "perduts":     [r["perduts"]     for r in registros],
        "punts":       [r["punts"]       for r in registros]
    }
````

## Análisis de negocio

- `equipo_mas_goleador(registros)`: devuelve el equipo con más goles a favor.
- `equipo_mas_goleado(registros)`: devuelve el equipo con más goles en contra.
- `mostrar_resultados(tabla, registros)`: imprime la clasificación y las respuestas.

````python
def equipo_mas_goleador(registros):

    """Devuelve el equipo con más goles a favor."""

    return max(registros, key=lambda r: r["gols_favor"])
````

````python
def equipo_mas_goleado(registros):

    """Devuelve el equipo con más goles en contra."""

    return max(registros, key=lambda r: r["gols_contra"])


````

````python
def mostrar_resultados(tabla, registros):

    """Muestra la tabla de clasificación y las respuestas de negocio."""

    print("\n" + "=" * 90)
    print("CLASIFICACIÓN FINAL")
    print("=" * 90)
    print(f"{'Pos':<4} {'Equipo':<20} {'GF':<5} {'GC':<5} {'G':<4} "
          f"{'E':<4} {'P':<4} {'Pts':<4}")
    print("-" * 90)
    for i, equipo in enumerate(tabla["equip"], start=1):
        idx = i - 1
        print(f"{i:<4} {equipo:<20} {tabla['gols_favor'][idx]:<5} "
              f"{tabla['gols_contra'][idx]:<5} {tabla['guanyats'][idx]:<4} "
              f"{tabla['empatats'][idx]:<4} {tabla['perduts'][idx]:<4} "
              f"{tabla['punts'][idx]:<4}")
    print("=" * 90)

    goleador = equipo_mas_goleador(registros)
    goleado  = equipo_mas_goleado(registros)

    print(f"\nEquipo más goleador: {goleador['equip']} "
          f"({goleador['gols_favor']} goles a favor)")
    print(f"Equipo más goleado:  {goleado['equip']} "
          f"({goleado['gols_contra']} goles en contra)")

    print("\nClasificación final (top 3):")
    for i, equipo in enumerate(tabla["equip"][:3], start=1):
        print(f"  {i}. {equipo} ({tabla['punts'][i-1]} pts)")
    print()
````

## Función principal y ejecución

`main_deportes()` orquesta todo el flujo: leer → procesar → ordenar → construir tabla → mostrar.

````python
def main_deportes(ruta="historic partits.txt"):

    """
    Función orquestadora: lee, procesa, ordena y muestra los resultados.
    """

    lineas = leer_archivo(ruta)
    estadisticas = procesar_partidos(lineas)
    registros = ordenar_por_puntos(estadisticas)
    tabla = construir_tabla(registros)
    mostrar_resultados(tabla, registros)


# Ejecución
main_deportes("historic partits.txt")
````
