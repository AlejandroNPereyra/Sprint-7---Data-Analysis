# EJERCICIO 1: Selección de Personal



Crear una herramienta en Python que actúe como formulario de inscripción y sistema de clasificación inicial para una oferta laboral interna.

**Reglas:**
- Solo se aceptan las **10 primeras personas válidas** que se inscriban.
- Cada candidato recibe una puntuación según experiencia, catalán, Excel y titulación.
- Se clasifica en: Prioridad Alta, Candidato Adecuado o No Prioritario.
- Al llegar a 10, se genera un informe final ordenado por prioridad y puntuación.

# Objetivo

Recoger datos de candidatos, validarlos, calcular su puntuación, clasificarlos y generar un informe final ordenado.

# Enfoque

1. **Recoger** los datos del candidato por teclado.
2. **Validar** que todos los campos son correctos.
3. **Calcular** la puntuación: experiencia × 1.5 + catalán + Excel + titulación.
4. **Clasificar** según los umbrales definidos.
5. **Mostrar** el resultado individual.
6. **Generar** un informe final ordenado por prioridad y puntuación.

# Restricción

Solo cuentan como inscritos los candidatos **válidos**. Un candidato inválido no ocupa plaza.

# Reglas de puntuación

| Concepto | Valor |
|---|---|
| Años de experiencia | × 1.5 |
| Catalán | 0–3 |
| Excel | 0–3 |
| Titulación | 0–3 |

**Fórmula:** puntuación = experiencia × 1.5 + catalán + excel + titulación

# Umbrales de clasificación

| Puntuación | Clasificación |
|---|---|
| ≥ 12 | Prioridad Alta |
| 8 ≤ puntuación < 12 | Candidato Adecuado |
| < 8 | No Prioritario |

# obtener_datos():

Pedir por teclado el nombre y apellidos y limpiar espacios sobrantes.

Pedir por teclado el teléfono de contacto y limpiar espacios sobrantes.

Pedir por teclado los años de experiencia y limpiar espacios sobrantes.

Pedir por teclado el nivel de catalán y limpiar espacios sobrantes.

Pedir por teclado el dominio de Excel y limpiar espacios sobrantes.

Pedir por teclado el nivel académico y limpiar espacios sobrantes.

Crear un diccionario con una clave por cada dato recogido.

Guardar cada respuesta en su clave correspondiente.

Devolver el diccionario.

```python
def obtener_datos():

    """
    Pide por teclado los datos del candidato.
    Devuelve un diccionario con todos los valores como str
    """

    nombre = input("Nombre y apellido: ").strip()
    telefono = input("Teléfono de contacto: ").strip()
    experiencia = input("Años de experiencia en la empresa: ").strip()
    catalan = input("Nivel de catalán (0-3): ").strip()
    excel = input("Dominio de Excel (0-3): ").strip()
    titulo = input("Nivel académico (0-3): ").strip()

    return {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulo": titulo,
    }
```

## validar_datos(datos):

Comprobar que el nombre no esté vacío; si lo está, avisar del error y devolver None.

Comprobar que el teléfono tenga nueve caracteres y que todos sean dígitos; si no, avisar del error y devolver None.

Intentar convertir la experiencia a un número entero; si no es posible, avisar del error y devolver None.

Comprobar que la experiencia sea mayor o igual que cero; si no, avisar del error y devolver None.

Para cada uno de los campos catalán, Excel y título:

    Intentar convertir el valor a número entero; si no es posible, avisar del error y devolver None.

    Comprobar que el valor esté entre cero y tres; si no, avisar del error y devolver None.

    Guardar el valor ya convertido.

Si todas las comprobaciones pasaron, construir y devolver un diccionario nuevo con los valores numéricos ya convertidos a int.

```python
def validar_datos(datos):

    """
    Valida cada campo y, si todo es correcto, devuelve un NUEVO
    diccionario con los valores numéricos ya convertidos a int.
    Si algo falla, imprime el error y devuelve None.
    """

    # Nombre
    if not datos["nombre"]:
        print("Error: el nombre no puede estar vacío.")
        return None

    # Teléfono
    if not (datos["telefono"].isdigit() and len(datos["telefono"]) == 9):
        print("Error: el teléfono debe tener 9 dígitos.")
        return None

    # Experiencia
    try:
        experiencia = int(datos["experiencia"])
    except ValueError:
        print("Error: los años de experiencia deben ser un entero >= 0.")
        return None
    if experiencia < 0:
        print("Error: los años de experiencia deben ser un entero >= 0.")
        return None

    # Niveles 0-3
    convertidos = {}
    for campo in ["catalan", "excel", "titulo"]:
        try:
            valor = int(datos[campo])
        except ValueError:
            print(f"Error: {campo} debe ser un entero entre 0 y 3.")
            return None
        if not 0 <= valor <= 3:
            print(f"Error: {campo} debe ser un entero entre 0 y 3.")
            return None
        convertidos[campo] = valor

    return {
        "nombre": datos["nombre"],
        "telefono": datos["telefono"],
        "experiencia": experiencia,
        "catalan": convertidos["catalan"],
        "excel": convertidos["excel"],
        "titulo": convertidos["titulo"],
    }
```

## calcular_puntuacion(datos):

Multiplicar los años de experiencia por uno coma cinco.

Sumar el nivel de catalán.

Sumar el dominio de Excel.

Sumar el nivel académico.

Devolver la puntuación total.

```python
def calcular_puntuacion(datos):

    """
    Aplica la fórmula del enunciado.
    Devuelve un diccionario con valores numéricos.
    """

    return (
        datos["experiencia"] * 1.5
        + datos["catalan"]
        + datos["excel"]
        + datos["titulo"]
    )
```

## clasificar_candidato(puntuacion):

Dentro de una condición.

Si la puntuación es mayor o igual que doce, devolver "Prioridad Alta".

Si no, si la puntuación es mayor o igual que ocho, devolver "Candidato Adecuado".

Si la puntuación no está dentro de lo anterior, devolver "No Prioritario".

```python
def clasificar_candidato(puntuacion):

    """
    Traduce la puntuación numérica a una etiqueta.
    """

    if puntuacion >= 12:
        return "Prioridad Alta"
    elif puntuacion >= 8:
        return "Candidato Adecuado"
    else:
        return "No Prioritario"
```

# mostrar_individual(datos, puntuacion, clasificacion):

Imprimir un separador de inicio.

Imprimir el nombre del candidato.

Imprimir el teléfono del candidato.

Imprimir la puntuación obtenida.

Imprimir la clasificación asignada.

Imprimir un separador de cierre.

```python
def mostrar_individual(datos, puntuacion, clasificacion):

    """
    Imprime por pantalla los datos del candidato una vez registrado.
    """

    print("\n--- Candidato registrado ---")
    print(f"Nombre: {datos['nombre']}")
    print(f"Teléfono: {datos['telefono']}")
    print(f"Puntuación: {puntuacion}")
    print(f"Clasificación: {clasificacion}")
    print("----------------------------\n")
```

# mostrar_informe(candidatos):

Definir un orden de prioridad: Alta primero, Adecuado después, No Prioritario al final.

Ordenar la lista de candidatos primero por ese orden de prioridad y, en caso de empate, por puntuación de mayor a menor.

Imprimir una cabecera con los títulos de las columnas.

Imprimir una línea separadora.

Para cada candidato en la lista ordenada, imprimir su nombre, teléfono, puntuación y clasificación.

Imprimir una línea de cierre.

```python
def mostrar_informe(candidatos):

    """
    Ordena por prioridad (Alta → Adecuado → No Prioritario) y,
    en caso de empate, por puntuación descendente.
    """

    orden_prioridad = {
        "Prioridad Alta": 0,
        "Candidato Adecuado": 1,
        "No Prioritario": 2,
    }

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda c: (orden_prioridad[c["clasificacion"]], -c["puntuacion"]),
    )

    print("\n========== INFORME FINAL ==========")
    print(f"{'Nombre':<25} {'Teléfono':<12} {'Puntuación':<10} {'Clasificación'}")
    print("-" * 70)
    for c in candidatos_ordenados:
        print(
            f"{c['nombre']:<25} {c['telefono']:<12} "
            f"{c['puntuacion']:<10} {c['clasificacion']}"
        )
    print("===================================\n")
```

# main_seleccion():

Inicializar una lista vacía de candidatos.

Definir el número máximo de candidatos como diez.

Mientras la lista tenga menos de diez candidatos:

    Mostrar en pantalla en qué número de inscripción va.
    Recoger los datos del candidato.
    Validar los datos.
    Si la validación falla, avisar de que debe reintentarlo y volver al inicio del bucle sin contar esa inscripción.
    Calcular la puntuación del candidato.
    Clasificar al candidato según su puntuación.
    Crear un diccionario con solo el nombre, el teléfono, la puntuación y la clasificación.
    Añadir ese diccionario a la lista de candidatos.
    Mostrar el resultado individual del candidato.
    
Al salir del bucle, generar el informe final con todos los candidatos válidos.

```python
def main_seleccion():

    """
    Orquesta el flujo completo:
    1. Inicializa lista y tope de candidatos.
    2. Repite hasta tener 10 válidos.
    3. Recoge, valida (convierte), calcula, clasifica, guarda y muestra.
    4. Al llegar a 10, genera el informe final.
    """

    candidatos = []
    MAX_CANDIDATOS = 10

    while len(candidatos) < MAX_CANDIDATOS:
        print(f"\nInscripción {len(candidatos) + 1} de {MAX_CANDIDATOS}")
        datos = obtener_datos()

        datos_validados = validar_datos(datos)
        if datos_validados is None:
            print("Vuelve a intentarlo.\n")
            continue  # el candidato inválido NO ocupa plaza

        puntuacion = calcular_puntuacion(datos_validados)
        clasificacion = clasificar_candidato(puntuacion)

        candidato = {
            "nombre": datos_validados["nombre"],
            "telefono": datos_validados["telefono"],
            "puntuacion": puntuacion,
            "clasificacion": clasificacion,
        }
        candidatos.append(candidato)

        mostrar_individual(datos_validados, puntuacion, clasificacion)

    mostrar_informe(candidatos)
```

## Ejecución

```python
main_seleccion()
```

- El informe final ordena primero por prioridad y, en caso de empate, por puntuación descendente.
